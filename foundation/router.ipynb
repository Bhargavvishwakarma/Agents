```python
import os
import json
from dotenv import load_dotenv
from openai import OpenAI
```

```python
load_dotenv(override=True)
```

```python
google_api_key = os.getenv('GOOGLE_API_KEY')
groq_api_key = os.getenv('GROQ_API_KEY')

if google_api_key:
    print("Gemini")
if groq_api_key:
    print("groq_api_key")
```

```python
from IPython.display import display, Markdown
gemini = OpenAI(api_key=google_api_key, base_url="https://generativelanguage.googleapis.com/v1beta/openai/")
groq = OpenAI(api_key=groq_api_key, base_url="https://api.groq.com/openai/v1")

def route_task(prompt):
    prompt_lower = prompt.lower()

    if "code" in prompt_lower or "python" in prompt_lower:
        return "groq"
    elif  "explain" in prompt_lower or "deep" in prompt_lower:
        return "gemini" 
    else:
        return "groq"

def call_groq(prompt):
    print(f"groq {prompt}")
    return groq.chat.completions.create(
        model= "openai/gpt-oss-120b",
        messages=[{
            "role": "user",
            "content" : prompt
        }]
    ).choices[0].message.content

def call_gemini(prompt):
    return gemini.chat.completions.create(
        model= "gemini-flash-latest",
        messages=[{
            "role": "user",
            "content" : prompt
        }]
    ).choices[0].message.content

def agent(prompt):
    route = route_task(prompt)
    if route == "gemini":
        print("Routed to Gemini")
        return call_gemini(prompt)
    else:
        print("Routed to Groq")
        return call_groq(prompt)


user_input = input("Enter your prompt: ")
result = agent(user_input)
display(Markdown(result))


```

```python
from IPython.display import display, Markdown
gemini = OpenAI(api_key=google_api_key, base_url="https://generativelanguage.googleapis.com/v1beta/openai/")
groq = OpenAI(api_key=groq_api_key, base_url="https://api.groq.com/openai/v1")

def intelligent_router(prompt):
    router_prompt = f"""
    Classify this task into one of:
    - coding
    - reasoning
    - simple

    Task: {prompt}

    Only return the label.
    """

    decision = call_groq(router_prompt).lower()

    if "coding" in decision:
        return "groq"
    elif "reasoning" in decision:
        return "gemini"
    else:
        return "groq"

def call_groq(prompt):
    print(f"groq {prompt}")
    return groq.chat.completions.create(
        model= "openai/gpt-oss-120b",
        messages=[{
            "role": "user",
            "content" : prompt
        }]
    ).choices[0].message.content

def call_gemini(prompt):
    return gemini.chat.completions.create(
        model= "gemini-flash-latest",
        messages=[{
            "role": "user",
            "content" : prompt
        }]
    ).choices[0].message.content

def agent(prompt):
    route = route_task(prompt)
    if route == "gemini":
        print("Routed to Gemini")
        return call_gemini(prompt)
    else:
        print("Routed to Groq")
        return call_groq(prompt)


user_input = input("Enter your prompt: ")
result = agent(user_input)
display(Markdown(result))


```
